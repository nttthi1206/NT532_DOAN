```python
import numpy as np
import pandas as pd

# Đường dẫn đến file vừa giải nén
data_path = "../data/raw/Sleep_health_and_lifestyle_dataset.csv"
# (Nếu file CSV để cùng cấp với notebook thì chỉ cần: "Sleep_health_and_lifestyle_dataset.csv")

df = pd.read_csv(data_path)

# Hiển thị 5 dòng đầu tiên
df.head()
```

```python
# Xem các kiểu dữ liệu, số lượng bản ghi
df.info()
df["Sleep Disorder"] = df["Sleep Disorder"].fillna("None")

# Kiểm tra xem có cột nào bị thiếu dữ liệu (NaN) hay không
print("--- Số lượng giá trị thiếu (NaN) trên từng cột ---")
print(df.isnull().sum())
```

```python
# 1. Chuẩn hóa nhãn 'Quality of Sleep' (1-10) thành Sleep Quality Score (0-100)
df["sleep_quality_score"] = df["Quality of Sleep"] * 10

# 2. Chọn tập đặc trưng chính phù hợp với dữ liệu cảm biến
selected_features = [
    "Heart Rate",
    "Physical Activity Level",
    "Daily Steps",
    "Sleep Duration",
    "Stress Level",
    "sleep_quality_score",
]

df_selected = df[selected_features].copy()
df_selected.head()
```

```python
df_selected.to_csv("../data/processed/sleep_cleaned.csv", index=False)
print("Đã lưu dữ liệu làm sạch thành công!")
import json

sample_data = {
    "device_id": "simulated_sleep_01",
    "timestamp": "2026-10-21T22:30:00+07:00",
    "heart_rate_bpm": int(df.loc[0, "Heart Rate"]),
    "movement": round(df.loc[0, "Physical Activity Level"] / 100.0, 2),
    "temperature_c": 26.5,  # giả lập nhiệt độ phòng ngủ bình thường
    "humidity_percent": 60,  # giả lập độ ẩm phòng
}

print("Mẫu dữ liệu MQTT Replay:")
print(json.dumps(sample_data, indent=2))
```
